## Part 2 · Named constants and formatted output

A **named constant** holds a value that should never change. Python has no `const` keyword,
so we write the name in `ALL_CAPS` as a promise to other programmers.

### GUIDED EXAMPLE 2.1 — constants beat magic numbers

Compare the two versions. Both compute the same answer; only one explains itself.

````python
# Version A - magic numbers
total_a = 40.00 + 40.00 * 0.0925
print(total_a)

# Version B - named constant
TAX_RATE = 0.0925
subtotal = 40.00
tax      = subtotal * TAX_RATE
total_b  = subtotal + tax
print(f"Subtotal: ${subtotal:.2f}")
print(f"Tax:      ${tax:.2f}")
print(f"Total:    ${total_b:.2f}")
````

### GUIDED EXAMPLE 2.2 — f-string format specifiers

Put an `f` before the quote, then write any expression inside `{ }`.
A colon introduces a **format specifier**.

````python
value = 1234.56789

print(f"Plain:        {value}")
print(f"2 decimals:   {value:.2f}")
print(f"0 decimals:   {value:.0f}")
print(f"With commas:  {value:,.2f}")
print(f"Right in 12:  |{value:>12.2f}|")
print(f"Left in 12:   |{value:<12.2f}|")
print(f"As percent:   {0.0925:.1%}")
````

## Part 4 · Type conversion in detail  

### GUIDED EXAMPLE 4.1 — what each conversion does

Note that `int()` **truncates** a float — it chops the decimals off, it does not round.

````python
print(int("42"))        # text of a whole number -> 42
print(int(9.87))        # TRUNCATES -> 9   (not 10!)
print(round(9.87))      # rounds     -> 10
print(float("3.5"))     # -> 3.5
print(float(7))         # -> 7.0
print(str(42) + "!")    # number -> text, so + concatenates
print(int(True))        # -> 1
print(bool(0), bool(""), bool("hi"))   # falsy, falsy, truthy
````
